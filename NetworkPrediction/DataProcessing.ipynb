```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
dataframe = pd.read_csv(r'D:\Main project\dataset\data.csv')
```

```python
cat_columns = ["Source.IP", "Destination.IP", "ProtocolName"]
label_encoders = {}
for col in cat_columns:
    le = LabelEncoder()
    dataframe[col] = le.fit_transform(dataframe[col])
    label_encoders[col] = le
```

```python
numeric_features = [
    "Flow.Duration", "Total.Fwd.Packets", "Total.Backward.Packets",
    "Total.Length.of.Fwd.Packets", "Total.Length.of.Bwd.Packets",
    "Flow.Bytes.s", "Flow.Packets.s", "Fwd.Packet.Length.Mean",
    "Bwd.Packet.Length.Mean", "Packet.Length.Mean"
]
```

```python
scaler = MinMaxScaler()
dataframe[numeric_features] = scaler.fit_transform(dataframe[numeric_features])
```

```python
pca = PCA(n_components=3)  # Reduce to 3 principal components
pca_features = pca.fit_transform(dataframe[numeric_features])
pca_df = pd.DataFrame(pca_features, columns=["PCA1", "PCA2", "PCA3"])
```

```python
final_df = pd.concat([dataframe[cat_columns], pca_df], axis=1)
```

```python
train_data, test_data = train_test_split(final_df, test_size=0.2, random_state=42, shuffle=True)
```

```python
train_data.to_csv(r'D:\Main project\dataset\train_data.csv', index=False)
test_data.to_csv(r'D:\Main project\dataset\test_data.csv', index=False)

print("✅ Data Preprocessing Completed Successfully!")
```
